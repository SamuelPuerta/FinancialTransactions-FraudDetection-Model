## 3. Estado del arte

### 3.1 Cómo se hizo la búsqueda

Primero se buscaron trabajos que usaran el mismo dataset de Kaggle, con su nombre, los nombres de sus variables (`risk_score`, `is_fraud`, `account_type`) y sus cifras (10.000 transacciones, 2.142 usuarios). No se encontró ningún artículo que lo use, lo cual es esperable porque es un conjunto sintético muy reciente.

Por eso se eligieron artículos que resuelven **el mismo problema con otros datos**: detectar transacciones fraudulentas como un problema de clasificación con clases muy desbalanceadas. Se pidieron tres condiciones: que fueran artículos de revista (IEEE o Elsevier), que fueran experimentales (no revisiones) y que explicaran con claridad cómo validaron sus modelos. Quedaron los cinco trabajos de la Tabla 1.

### 3.3 Qué hizo cada trabajo

**Dal Pozzolo et al. (2018) [1].** Plantean la detección de fraude como la vive un banco: las transacciones llegan todos los días y la etiqueta de fraude solo se conoce días después, cuando un investigador revisa el caso. Proponen entrenar dos modelos de Random Forest por separado, uno con los casos revisados recientemente y otro con los casos antiguos, y promediar sus predicciones. Validan de forma temporal, entrenando con los días pasados y evaluando el día siguiente. Con su método, el 77 % de las 100 alertas diarias son fraudes reales, frente al 61 % de un solo modelo entrenado con todos los datos. También muestran que el AUC puede preferir un modelo que genera peores alertas.

**Correa Bahnsen et al. (2016) [2].** Estudian cómo construir mejores variables. Además de los datos de cada transacción, crean variables que resumen el historial del cliente (por ejemplo, cuánto ha gastado en los últimos días) y variables periódicas, modeladas con la distribución de von Mises, que capturan a qué hora suele comprar. Comparan modelos clásicos y versiones sensibles al costo, que tienen en cuenta que no detectar un fraude cuesta el monto de la transacción. Usan una partición 50/25/25 y miden el ahorro económico que logra cada modelo. Las variables periódicas aumentan el ahorro promedio en un 13 %, y el mejor modelo es un árbol de decisión sensible al costo.

**Dal Pozzolo et al. (2014) [3].** Comparan Random Forest, redes neuronales y SVM con tres formas de manejar el desbalance: submuestreo, SMOTE y EasyEnsemble (varios submuestreos combinados). Usan 2,2 millones de transacciones reales y validan de forma temporal, entrenando con 30 a 90 días y prediciendo el día siguiente. Insisten en que la exactitud engaña en problemas desbalanceados y usan Average Precision, AUC y la precisión en las primeras alertas. Random Forest fue el mejor modelo, SMOTE y EasyEnsemble superaron al submuestreo, y reentrenar el modelo a diario fue mejor que dejarlo fijo.

**Carcillo et al. (2021) [4].** Combinan aprendizaje no supervisado y supervisado: calculan puntajes de "qué tan rara es la transacción" con técnicas como Isolation Forest o PCA y los añaden como variables a un Random Forest balanceado. Usan 76 millones de transacciones reales con un corte temporal entre entrenamiento y prueba. Solo los puntajes calculados por grupos de clientes parecidos mejoran un poco la PR-AUC, y añadir muchos puntajes a la vez produce sobreajuste.

**Makki et al. (2019) [5].** Es el trabajo más parecido a este proyecto, porque compara casi los mismos modelos: regresión logística, KNN, SVM, red neuronal y árboles de decisión. Primero los compara sin tratar el desbalance y luego les aplica sobremuestreo y aprendizaje sensible al costo, con una partición 70/30. La regresión logística obtuvo la mejor PR-AUC (0,66) y KNN la peor (0,29). Balancear los datos subió la sensibilidad a cerca del 65 %, pero a costa de muchas falsas alarmas.

### 3.4 Métricas que no se vieron en clase

**PR-AUC o Average Precision (AP)** [3], [4], [5]. Resume en un número la curva precisión–exhaustividad. Si se ordenan las transacciones de mayor a menor puntaje y se llaman $P_k$ y $R_k$ a la precisión y la exhaustividad al marcar como fraude las $k$ primeras:

$$\mathrm{AP}=\sum_{k=1}^{N}\left(R_k-R_{k-1}\right)P_k$$

Un modelo que asigna puntajes al azar obtiene una AP igual a la proporción de fraudes; en este proyecto, 0,0126.

**Precisión en las $k$ primeras alertas (Precision@k)** [1], [3], [4]. Es la proporción de fraudes reales entre las $k$ transacciones con mayor puntaje. Refleja que un banco solo puede revisar un número limitado de alertas por día:

$$\mathrm{Precision@}k=\frac{\text{fraudes entre las } k \text{ transacciones con mayor puntaje}}{k}$$

**Savings (ahorro)** [2]. Mide cuánto dinero ahorra el modelo frente a la mejor estrategia sin modelo (aprobar todo o bloquear todo). Si $y_i$ es la etiqueta real, $c_i$ la predicción, $\mathrm{Monto}_i$ el valor de la transacción y $C_a$ el costo de revisar una alerta:

$$\mathrm{Costo}=\sum_{i=1}^{N}\left[y_i\,(1-c_i)\,\mathrm{Monto}_i+c_i\,C_a\right],\qquad \mathrm{Savings}=\frac{\mathrm{Costo}_{\text{sin modelo}}-\mathrm{Costo}_{\text{modelo}}}{\mathrm{Costo}_{\text{sin modelo}}}$$

### 3.5 Qué aporta la literatura a este proyecto

- **Métricas.** Ningún trabajo usa la exactitud como criterio. Esto respalda usar la PR-AUC como métrica principal y reportar también la precisión en las alertas con mayor puntaje.
- **Desbalance.** Balancear los datos no siempre ayuda. En [3], SMOTE funcionó mejor que el submuestreo; en [5], balancear subió la sensibilidad pero disparó las falsas alarmas; y [2] ni siquiera lo usa. Por eso se compararán la ponderación de clases y SMOTE, aplicados siempre solo sobre los datos de entrenamiento.
- **Validación.** Los trabajos con fechas validan respetando el tiempo. Como este dataset tiene `timestamp`, además de la validación cruzada estratificada se hará una partición temporal.
- **Modelos.** Random Forest es el referente en [1]–[4]. Sin embargo, en [5] la regresión logística obtuvo la mejor PR-AUC y KNN la peor, lo que da una hipótesis concreta para contrastar con los modelos de este proyecto.
- **Diferencias con este proyecto.** Los trabajos revisados usan millones de transacciones reales y variables del historial del cliente, que según [2] son de las más útiles. Este dataset es pequeño (10.000 transacciones y 126 fraudes), sintético y, como se mostró en la sección 2, el historial del usuario no aporta señal. Además, se ha reportado que los mismos modelos rinden significativamente peor sobre datos simulados que sobre datos reales [6]. Por todo esto, se esperan resultados bastante más bajos que los de la literatura.
